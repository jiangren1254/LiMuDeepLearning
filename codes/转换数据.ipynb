```python
import random
import time

def generate_random_numbers(n, length = 1):
    if length <= 0 or n <= 0:
        raise ValueError("n 和 length 必须大于 0")
    start = 10**(length-1)  # 例如 length=3 时，最小值是100
    end = 10**length - 1  # 例如 length=3 时，最大值是999
    return [random.randint(start, end) for _ in range(n)]
n = 1000
a = generate_random_numbers(n)
b = generate_random_numbers(n)
c= generate_random_numbers(n)

```

```python
t = time.time()
for i in range(n):
    c[i] = a[i] + b[i]
print(time.time() - t)
```
